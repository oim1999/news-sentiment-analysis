# Stock market analysis of AAPL

### Stock market analysis helps to :-
Evaluate investment opportunities in specific stocks.
- Identify trends and patterns in the market.
- Make informed investment decisions based on data.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import warnings
warnings.filterwarnings('ignore')
```

```python
TICKER = 'AAPL'
df = pd.read_csv(f'{TICKER}.csv', parse_dates=['Date'])
df.set_index('Date', inplace=True)
df['20_SMA'] = df['Close'].rolling(window=20).mean()
df['50_SMA'] = df['Close'].rolling(window=50).mean()
```
